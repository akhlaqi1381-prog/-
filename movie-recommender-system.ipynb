```python
movies["genres_clean"] = movies["genres"].str.split("|")
```

```python
from sklearn.preprocessing import MultiLabelBinarizer
mlb=MultiLabelBinarizer()
genre_matrix=mlb.fit_transform(movies["genres_clean"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity_matrix=cosine_similarity(genre_matrix)
```

```python
movie_title = "Toy Story (1995)"
movie_index = movies[movies["title"] == movie_title].index[0]

print(movie_index)
```

```python
similarity_score=list(
    enumerate(similarity_matrix[movie_index]
))
#print(similarity_score)
similarity_score=sorted(
    similarity_score,key=lambda x: x[1],
    reverse=True)

Film=similarity_score[1:11]
for index, score in Film:
    print(
        movies.iloc[index]["title"],
        "→",
        round(score, 3)
    )
```

```python
#پیشنهاد فیلم به کاربر بر اساس سلیقه مشابه کاربر هایی که
#چند فیلم را دیدن و رای یکسانی به انها دادن
```

```python
#اابتدا باید جدولی بسازیم که
#سطر های ان کاربران و ستون های ان فیلم ها و داخل سلول ها
#نظرات کاربر به ان فیلم ها باشد
```

```python
user_movie_matrix=ratings.pivot_table(
    index="userId",
    columns="movieId",
    values="rating"
)
     
```

```python
from sklearn.metrics.pairwise import cosine_similarity

user_similarity = cosine_similarity(
    user_movie_matrix.fillna(0)
)

print(user_similarity)
```

```python
user_id = 1

user_index = user_movie_matrix.index.get_loc(user_id)

similar_users = list(
    enumerate(user_similarity[user_index])
)

similar_users = sorted(
    similar_users,
    key=lambda x: x[1],
    reverse=True
)

similar_users = similar_users[1:11]

for index, score in similar_users:
    similar_user_id = user_movie_matrix.index[index]
    
    print(
        "User:", similar_user_id,
        "→",
        round(score, 3)
    )
```

```python
#ابتدا این ۱۰ نفر را وارد یک لیست میکنیم
top_10_users = []

for index, similarity in similar_users:
    similar_user_id = user_movie_matrix.index[index]
    
    top_10_users.append(similar_user_id)

print(top_10_users)
```

```python
#فیلم هایی ک این ۱۰ نفر دیده اند را جمع میکنیم در یک لیست
candidate_movies = []
for user_id in top_10_users:
    user_movies = user_movie_matrix.loc[user_id].dropna().index
    candidate_movies.extend(user_movies)
```

```python
#فیلم هایی ک کاربر یک دیده را از اون لیست پاک میکنیم
user_1_movies = user_movie_matrix.loc[1].dropna().index
new_candidate_movies = []

for movie_id in candidate_movies:
    
    if movie_id not in user_1_movies:
        
        new_candidate_movies.append(movie_id)
print(new_candidate_movies)
```

```python
#حالا این فیلم هارا میشماریم و فیلم هایی ک بیشترین تکرار را
#دارند ۱۰ تای اول رو پیشنهاد میدهیم
movie_counts = pd.Series(new_candidate_movies).value_counts()
#print(movie_counts)
top_10=movie_counts[:10]
#print(top_10)
```

```python
for movie_id, count in top_10.items():

    title = movies.loc[
        movies["movieId"] == movie_id,
        "title"
    ].iloc[0]

    print(title, "→", count, "نفر")
```
